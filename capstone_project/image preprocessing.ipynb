```python
import cv2  # OpenCV 라이브러리 불러오기
import os   # 파일 및 폴더 경로 작업용 모듈
import ffmpeg  # ffmpeg-python 라이브러리
```

```python
def fix_rotation(input_path, temp_path):
    """
    MOV 파일의 회전 정보를 자동으로 수정하는 함수
    :param input_path: 원본 MOV 파일 경로
    :param temp_path: 회전 정보를 수정한 임시 동영상 파일 경로
    """
    try:
        # ffmpeg를 사용하여 회전 정보 수정 후 임시 파일로 저장
        ffmpeg.input(input_path).output(temp_path, vcodec='libx264', acodec='aac').run()
        print(f"✅ 회전 정보 처리 완료: {temp_path}")
    except ffmpeg.Error as e:
        print(f"❌ 오류 발생: {e}")
```

```python
def adjust_video_length(input_path, output_path, target_duration=10, target_fps=30, target_size=(1920, 1080)):
    # 임시 동영상 경로 (회전 수정된 파일)
    temp_path = "temp_fixed_video.mp4"

    # MOV 파일에 회전 정보가 있다면 수정
    if input_path.lower().endswith(".mov"):
        fix_rotation(input_path, temp_path)  # 회전 정보 수정
        input_path = temp_path  # 수정된 동영상 경로로 변경

    cap = cv2.VideoCapture(input_path)  # 동영상 파일 열기

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))  # 원본 총 프레임 수
    original_fps = cap.get(cv2.CAP_PROP_FPS)               # 원본 프레임 수
    target_frames = int(target_fps * target_duration)      # 목표 프레임 수 계산

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 출력 동영상의 코덱 설정
    out = cv2.VideoWriter(output_path, fourcc, target_fps, target_size)  # 출력 동영상 설정 (fps, 해상도)

    frames = []  # 프레임 저장 리스트
    success, frame = cap.read()
    while success:
        resized_frame = cv2.resize(frame, target_size)  # 프레임을 지정된 해상도로 리사이즈
        frames.append(resized_frame)
        success, frame = cap.read()

    cap.release()  # 원본 동영상 닫기

    # 프레임 재배치 (프레임 수 맞추기)
    if len(frames) >= target_frames:
        frames = frames[:target_frames]  # 프레임 수가 많으면 자르기
    else:
        last_frame = frames[-1] if frames else None
        while len(frames) < target_frames:
            frames.append(last_frame)   # 부족하면 마지막 프레임 반복

    for frame in frames:
        out.write(frame)  # 프레임 출력

    out.release()  # 결과 저장
    print(f"✅ {os.path.basename(input_path)} → 완료")

    # 임시 파일 삭제
    if os.path.exists(temp_path):
        os.remove(temp_path)
```

```python
def process_all_videos_in_folder(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)  # 출력 폴더가 없으면 생성
    for filename in os.listdir(input_folder):  # 입력 폴더 내 파일들 순회
        if filename.lower().endswith((".mp4", ".mov")):  # .mp4 또는 .mov만 처리
            input_path = os.path.join(input_folder, filename)   # 입력 파일 경로
            output_path = os.path.join(output_folder, filename) # 출력 파일 경로
            adjust_video_length(input_path, output_path)        # 변환 실행
```

```python
import cv2  # OpenCV 라이브러리 불러오기
import os   # 파일 및 폴더 경로 작업용 모듈
import ffmpeg  # ffmpeg-python 라이브러리

def fix_rotation(input_path, temp_path):
    """
    MOV 파일의 회전 정보를 자동으로 수정하는 함수
    :param input_path: 원본 MOV 파일 경로
    :param temp_path: 회전 정보를 수정한 임시 동영상 파일 경로
    """
    try:
        # ffmpeg를 사용하여 회전 정보 수정 후 임시 파일로 저장
        ffmpeg.input(input_path).output(temp_path, vcodec='libx264', acodec='aac').run()
        print(f"✅ 회전 정보 처리 완료: {temp_path}")
    except ffmpeg.Error as e:
        print(f"❌ 오류 발생: {e}")

def adjust_video_length(input_path, output_path, target_duration=10, target_fps=30, target_size=(1920, 1080)):
    # 임시 동영상 경로 (회전 수정된 파일)
    temp_path = "temp_fixed_video.mp4"

    # MOV 파일에 회전 정보가 있다면 수정
    if input_path.lower().endswith(".mov"):
        fix_rotation(input_path, temp_path)  # 회전 정보 수정
        input_path = temp_path  # 수정된 동영상 경로로 변경

    cap = cv2.VideoCapture(input_path)  # 동영상 파일 열기

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))  # 원본 총 프레임 수
    original_fps = cap.get(cv2.CAP_PROP_FPS)               # 원본 프레임 수
    target_frames = int(target_fps * target_duration)      # 목표 프레임 수 계산

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 출력 동영상의 코덱 설정
    out = cv2.VideoWriter(output_path, fourcc, target_fps, target_size)  # 출력 동영상 설정 (fps, 해상도)

    frames = []  # 프레임 저장 리스트
    success, frame = cap.read()
    while success:
        resized_frame = cv2.resize(frame, target_size)  # 프레임을 지정된 해상도로 리사이즈
        frames.append(resized_frame)
        success, frame = cap.read()

    cap.release()  # 원본 동영상 닫기

    # 프레임 재배치 (프레임 수 맞추기)
    if len(frames) >= target_frames:
        frames = frames[:target_frames]  # 프레임 수가 많으면 자르기
    else:
        last_frame = frames[-1] if frames else None
        while len(frames) < target_frames:
            frames.append(last_frame)   # 부족하면 마지막 프레임 반복

    for frame in frames:
        out.write(frame)  # 프레임 출력

    out.release()  # 결과 저장
    print(f"✅ {os.path.basename(input_path)} → 완료")

    # 임시 파일 삭제
    if os.path.exists(temp_path):
        os.remove(temp_path)

def process_all_videos_in_folder(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)  # 출력 폴더가 없으면 생성
    for filename in os.listdir(input_folder):  # 입력 폴더 내 파일들 순회
        if filename.lower().endswith((".mp4", ".mov")):  # .mp4 또는 .mov만 처리
            input_path = os.path.join(input_folder, filename)   # 입력 파일 경로
            output_path = os.path.join(output_folder, filename) # 출력 파일 경로
            adjust_video_length(input_path, output_path)        # 변환 실행

# 사용 예시
input_folder = '/content/drive/MyDrive/dataset/유리가깨져서발을다쳤어요/'    # 원본 동영상 폴더 경로
output_folder = '/content/drive/MyDrive/유리가깨져서발을다쳤어요/'  # 처리된 동영상 저장 폴더 경로
process_all_videos_in_folder(input_folder, output_folder)
```
